import array as arr

```python
a=arr.array('i',[11,22,33,44,55,66,77])
```

```python
# Forward array printing with Positive indexs.

for i in range(0,len(a)):
    print(a[i],end=' ')
```

```python
# Reverse array printing with Positive indexs.

for i in range(len(a)-1,-1,-1):
    print(a[i],end=' ')
```

```python
# Forward Array priting with Negative indexs.

for i in range(-len(a),0):
    print(a[i],end=' ')
```

```python
# Reverse array printing with Negative indexs.

for i in range(-1,-len(a)-1,-1):
    print(a[i],end=' ')
```
